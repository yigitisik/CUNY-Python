```python
#Manipulating Images as ndarrays
img = datasets.face()
plt.imshow(img)
```

```python
#What is the data type of img? 
print(type(img))
#Also, what is the shape of img and how many dimensions does it have? 
print(f"shape of img: {img.shape} and dimension count: {img.ndim}")
```

```python
#Now can you try and convert the image to black and white? All you need need to do is use a formula.
#Y_linear is what we're after - our black and white image. However, this formula only works if our rgb vals
#are between 0 and 1 - namely in sRGB format. Currently the values in our img range from 0 to 255.
sRGB = img / 255
# Multiply the sRGB array by the grey_vals array (provided) to convert the image to grayscale.
grey_vals = np.array([0.2126, 0.7152, 0.0722])
img_gray = sRGB @ grey_vals
#could also do img_gray = np.matmul(sRGB, img_gray)
# Finally use Matplotlib's .imshow() with the colormap parameter set to gray cmap=gray to display the result.
plt.imshow(img_gray, cmap='gray')
```

```python
#See if you can change the values in the ndarray so that:

#1) You flip the grayscale image upside down:
img_flipped = np.flip(img_gray)
plt.imshow(img_flipped, cmap='gray')
```

```python
#2) Rotate the colour image:
img_rotate = np.rot90(img)
plt.imshow(img_rotate)
```

```python
#3) Invert (i.e., solarize) the colour image. 
#To do this you need to convert all the pixels to their "opposite" value, so black (0) becomes white (255).
solar_img = 255 - img
plt.imshow(solar_img)
```

```python
def logger(func):
    def wrapper():
        print(f"Function {func.__name__} is being called")
        func()
        print(f"Function {func.__name__} has finished execution")
    return wrapper

@logger
def say_hello():
    print("Hello!")
    
say_hello()
```
